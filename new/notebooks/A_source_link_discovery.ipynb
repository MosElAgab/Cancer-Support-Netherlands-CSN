# A. Source Link Discovery

This notebook crawls trusted source websites and creates the raw `discovered_links_df`.

Each row is a link observation. Translation, enrichment, filtering, scoring, and classification are handled in later notebooks.

## 1. Imports

```python
from pathlib import Path
import sys

import pandas as pd

sys.path.append("..")

from src.crawling import crawl_site
```

## 2. Output Path

```python
RAW_DATA_DIR = Path("../data/raw")
RAW_DATA_DIR.mkdir(parents=True, exist_ok=True)

DISCOVERED_LINKS_PATH = RAW_DATA_DIR / "discovered_links_raw.csv"
```

## 3. Create Empty Discovered Links DataFrame

```python
discovered_links_df = pd.DataFrame()
discovered_links_df
```

## 4. Crawl kanker.nl

```python
start_url = "https://www.kanker.nl/"
base_url = "https://www.kanker.nl"
base_domain = "www.kanker.nl"
source_organisation = "kanker.nl"
blocked_sections = ["user", "overig"]
headers = {"User-Agent": "Mozilla/5.0"}

kanker_links_df = crawl_site(
    start_url=start_url,
    base_url=base_url,
    base_domain=base_domain,
    source_organisation=source_organisation,
    blocked_sections=blocked_sections,
    headers=headers,
    max_depth=4,
    max_pages=50,
)

kanker_links_df.head()
```

```python
discovered_links_df = pd.concat(
    [discovered_links_df, kanker_links_df],
    ignore_index=True,
)

discovered_links_df["source_organisation"].value_counts()
```
